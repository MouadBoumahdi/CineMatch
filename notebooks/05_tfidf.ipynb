## Imports and setup

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
```

## Read summaries

```python
movies = pd.read_csv("data/processed/movies_features.csv")

texts = movies["overview"].fillna("")

print("Shape:", texts.shape)
print("Missing values:", texts.isna().sum())
print(texts.head(3))
```

## Convert summaries

```python
vectorizer = TfidfVectorizer(max_features=1000)

X = vectorizer.fit_transform(texts)
```

## Inspect dimensions and terms

```python
print("Words:", vectorizer.get_feature_names_out()[:10])
print("Shape:",X.shape)
```
